Сортування вставками

Найкращий випадок: масив вже відсортований → кожен елемент порівнюється 1 раз → O(n)

Найгірший випадок: масив відсортований у зворотному порядку → кожен елемент проходить всі попередні → O(n²)

Середній випадок → O(n²)

Сортування бульбашкою

Найкращий випадок: відсортований масив якщо є оптимізація → O(n)

Найгірший випадок: зворотний порядок → O(n²)

Середній випадок → O(n²)

2.Реалізувати на Python алгоритм сортування методом бульбашки.

```python
def bubble_sort(arr):
    a = arr.copy()  
    n = len(a)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if not swapped:
            break

    return a
```

3.Дослідження реального часу виконання обох алгоритмів на Python

```python
import time
import random

def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr
n_values = [5, 10, 50, 100, 500, 1000, 2000, 3000, 4000, 5000, 10000, 20000, 50000, 100000]

time_insertion = []
time_bubble = []

num_trials = 3

print("Початок вимірювання часу виконання...")

for n in n_values:
    total_time_insert = 0
    for _ in range(num_trials):
        arr = [random.randint(0, 100000) for _ in range(n)]
        
        start_time = time.time()
        insertion_sort(arr.copy())
        end_time = time.time()
        total_time_insert += (end_time - start_time)
        
    time_insertion.append(total_time_insert / num_trials)
    
    total_time_bubble = 0
    for _ in range(num_trials):
        arr = [random.randint(0, 100000) for _ in range(n)]
        
        start_time = time.time()
        bubble_sort(arr.copy())
        end_time = time.time()
        total_time_bubble += (end_time - start_time)
        
    time_bubble.append(total_time_bubble / num_trials)
    
    print(f"Обробка n={n} завершена. Час Insertion: {time_insertion[-1]:.4f} с, Час Bubble: {time_bubble[-1]:.4f} с")

print("Вимірювання завершено.")
```

1.Що таке O-нотація і чим вона відрізняється від Ω -нотації?

O-нотація та Ω-нотація - це способи описати швидкість роботи алгоритму, O-нотація показує, наскільки довго може працювати алгоритм у найгіршому випадку, це максимальна оцінка часу. Ω -нотація показує, наскільки швидко принаймні буде працювати алгоритм, це мінімальна оцінка часу.

2.Яку часову складність має алгоритм сортування методом вставляння за найсприятливіших умов, коли масив відсортовано? Відповідь обґрунтувати.

У найсприятливіших умовах алгоритм вставляння має часову скаладність O(n). У сортуванні вставками кожен елемент порівнюється з попередніми, поки не знайде правильне місце. У загальному випадку - це може бути багато порівнянь і зсувів. Але у вже відсортованому масиві на кожній ітерації елемент key порівнюється лише один раз, умова key < a[j] одразу дає False, тому внутрішній цикл не виконується, зсувів немає. Отже, для кожного з n елементів виконується лише одна операція порівняння, без переміщень.
