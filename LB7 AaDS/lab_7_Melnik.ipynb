− Оцінити асимптотичну складність (в середньому і в найгіршому випадку) процедур search, insert і delete роботи з деревом.

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def insert(root, value):
    if root is None:
        return Node(value)

    if value < root.value:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)

    return root

def delete_subtree(root, key):
    if root is None:
        return None

    if root.value == key:
        return None

    root.left = delete_subtree(root.left, key)
    root.right = delete_subtree(root.right, key)

    return root

def inorder(root):
    if root is not None:
        inorder(root.left)
        print(root.value, end=" ")
        inorder(root.right)

root = None
for x in [8, 3, 10, 1, 6, 14, 4, 7, 13]:
    root = insert(root, x)

print("Дерево до видалення:")
inorder(root)
print()

root = delete_subtree(root, 3)

print("Дерево після видалення гілки 3:")
inorder(root)
print()
```

− Додати до нього елемент, який дорівнює вашому порядковому номеру у списку групи.

− Оцінити асимптотичну складність (у середньому і в найгіршому випадку) процедур search, insert і delete роботи з купою.

```python
import random
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def build_heap(arr):
    n = len(arr)

    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

def heap_insert(heap, value):
    heap.append(value)
    i = len(heap) - 1

    while i > 0:
        parent = (i - 1) // 2
        if heap[i] > heap[parent]:
            heap[i], heap[parent] = heap[parent], heap[i]
            i = parent
        else:
            break


def heap_extract_max(heap):
    if len(heap) == 0:
        return None

    maximum = heap[0]
    heap[0] = heap[-1]
    heap.pop()
    heapify(heap, len(heap), 0)

    return maximum

arr = [random.randint(1, 100) for _ in range(10)]
print("Рандомний масив:", arr)

build_heap(arr)
print("Купа:", arr)

heap_insert(arr, 13)
print("+13:", arr)

max_elem = heap_extract_max(arr)
print("Видалений максимальний елемент:", max_elem)
print("Купа після видалення максимуму:", arr)
```

3.Проведіть тестування геш-таблиці з різними типами даних (цілі числа, рядки, списки, словники, об’єкти):

– перевірити працездатність геш-таблиці з різними типами даних.

– виміряти час виконання основних операцій (пошук, вставка, видалення) для різних типів даних.

– порівняти результати для різних типів даних.

```python
import time

class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]

    def _hash(self, key):
        return hash(str(key)) % self.size

    def insert(self, key, value):
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))

    def search(self, key):
        index = self._hash(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        return None

    def delete(self, key):
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                del self.table[index][i]
                return True
        return False

    def __str__(self):
        return str(self.table)

ht = HashTable(size=20)

test_data = {
    "int": 42,
    "str": "hello",
    "list": [1, 2, 3],
    "dict": {"a": 1, "b": 2},
    "tuple": (5, 6),
}

results = []

for dtype, key in test_data.items():
    value = f"{dtype}_value"

    start = time.time()
    ht.insert(key, value)
    t_insert = time.time() - start

    start = time.time()
    v_found = ht.search(key)
    t_search = time.time() - start

    start = time.time()
    ht.delete(key)
    t_delete = time.time() - start

    results.append((dtype, t_insert, t_search, t_delete, v_found))

print(f"{'Тип':<10} {'Insert':<12} {'Search':<12} {'Delete':<12} {'Знайдено':<12}")
for r in results:
    print(f"{r[0]:<10} {r[1]:<12.6f} {r[2]:<12.6f} {r[3]:<12.6f} {str(r[4]):<12}")
```

1.Чим відрізняється структура бінарне дерево він бінарного дерева пошуку?

Основна відмінність полягає в структурі даних: бінарне дерево є загальною структурою, де кожен вузол має максимум два дочірні елементи, тоді як бінарне дерево пошуку (BST) має додаткову властивість: для будь-якого вузла всі ключі в його лівому піддереві менші за ключ самого вузла, а всі ключі в правому піддереві більші.

2.Чим відрізняється структура бінарне дерево від бінарної купи?

Основна відмінність між бінарним деревом і бінарною купою полягає в їхніх властивостях сортування та збалансованості: бінарне дерево (як-от бінарне дерево пошуку) відповідає правилу сортування, де для кожного вузла значення лівого піддерева менші, а правого — більші, тоді як бінарна купа відповідає властивості «купи», де батьківський вузол завжди менший (для min-купи) або більший (для max-купи) за своїх нащадків. Бінарне дерево може бути незбалансованим, а бінарна купа завжди є майже повною (на нижньому рівні всі вузли зліва). 

3.Які існують типи дерев? Опишіть їхні основні характеристики та переваги.

Бінарне дерево

Характеристики:

Кожен вузол має не більше двох дітей (ліва і права).

Може бути неповним або виродженим.

Не обов’язково впорядковане за значеннями.

Переваги:

Проста структура для зберігання ієрархічних даних.

Легко реалізується та обходиться.

Використовується у виразах, синтаксичних деревах, парсингу.

Бінарне дерево пошуку

Характеристики:

Підтип бінарного дерева.

Для кожного вузла: значення лівого піддерева ≤ вузол ≤ значення правого піддерева.

Дозволяє швидко шукати елементи.

Переваги:

Пошук, вставка, видалення — O(log n) для збалансованого дерева.

Використовується у словниках, базах даних, індексах.

Збалансовані дерева

Приклади:

AVL-дерево: балансування за висотою ±1

Red-Black дерево: балансування через кольори вузлів

B-дерево / B+ дерево: вузли з кількома ключами для зберігання на диску

Переваги:

Гарантовано O(log n) для пошуку, вставки, видалення.

Підходить для великих структур на диску (B-дерева).

N-арні дерева

Характеристики:

Кожен вузол може мати N дітей.

Використовується для зберігання ієрархій, наприклад: файлові системи.

Переваги:

Гнучкість у кількості дочірніх вузлів.

Зручне для представлення складних структур.

Дерева префіксів

Характеристики:

Використовуються для зберігання рядків.

Кожен вузол представляє символ.

Шлях від кореня до листа — слово або префікс.

Переваги:

Швидкий пошук префіксів і повних слів.

Застосування: автодоповнення, словники, пошук у рядках.

Купи

Характеристики:

Повне бінарне дерево.

Max-heap: кожен вузол ≥ дочірніх

Min-heap: кожен вузол ≤ дочірніх

Переваги:

Швидкий доступ до максимального/мінімального елементу: O(1).

Використовується у пріоритетних чергах, сортуванні heapsort.

Спеціалізовані дерева

Segment Tree (дерево відрізків) — для швидкого запиту і оновлення відрізків масиву.

Fenwick Tree (дерево Фенвіка) — для префіксних сум.

Suffix Tree — для рядкових алгоритмів (пошук підрядка).

Переваги:

Оптимізація запитів і оновлень.

Використання у алгоритмах на масивах та рядках.

5.Як організована купа? Опишіть алгоритми додавання та вилучення елементів з купи.

Купа - спеціалізована деревоподібна структура даних, в якій існують певні властивості впорядкованості: якщо В — вузол нащадок A — тоді ключ(A) ≥ ключ(B). З цього випливає, що елемент з найбільшим ключем завжди є кореневим вузлом.

Додавання

1. Додати елемент в кінець масиву.

2. Порівняти елемент із батьком:

3. Якщо елемент більший за батька → поміняти місцями.

Вилучення

1. Запам’ятати корінь як елемент для повернення.

2. Замінити корінь останнім елементом масиву (останній вузол) і видалити останній елемент.

3. Порівняти новий корінь з його дітьми:

4. Якщо один з дітей більший → поміняти місцями з більшим із них.

5. Повторювати крок 3, поки властивість купи не буде відновлена.

6. Повторювати крок 2, поки елемент не стане коренем або не буде меншим за батька.

Зберігання даних

Зберігаємо пару (key, value) у масиві за цим індексом.

Якщо знайдено → повертаємо значення.

Якщо немає → ключ відсутній у таблиці.

8.Опишіть методи вирішення колізій в хеш-таблицях. Які їхні переваги та недоліки?

Ланцюгове хешування 

Ідея:

1. Кожна комірка масиву таблиці містить список (ланцюжок) пар (key, value).

2. Якщо кілька ключів потрапляють на один індекс → вони зберігаються у списку.

Алгоритм:

1. Обчислити хеш ключа: index = h(key) % size

2. Додати (key, value) до списку у цій комірці.

3. Пошук: перевіряємо список у комірці на наявність ключа.

Переваги:

1. Простота реалізації.

2. Масив може бути меншого розміру.

3. Легко обробляти великі обсяги даних.

Недоліки:

1. У гіршому випадку (всі ключі в одному списку) пошук стає O(n).

2. Додатковий простір для списків.

Відкрите адресування

Ідея:

1. Усі елементи зберігаються безпосередньо в масиві.

2. Якщо індекс зайнятий → шукаємо наступну вільну комірку за певною схемою.

Схеми пробування:

1. Лінійне пробування

1. Якщо index зайнятий → перевіряємо index+1, index+2, ...

2. Плюси: проста реалізація

3. Мінуси: утворюються «кластеризації», погіршується ефективність

2. Квадратичне пробування

1. Перевіряємо комірки за квадратичною формулою: index + 1^2, index + 2^2 ...

2. Менше кластеризації, ніж у лінійного пробування

3. Може не знайти вільну комірку при великому заповненні

3. Подвійне хешування

1. Використовується друга хеш-функція для визначення кроку переміщення

2. Найменше кластеризації, висока ефективність

Переваги:

1. Не потребує додаткових списків → економія пам’яті

Проста реалізація для невеликих таблиць

Недоліки:

Таблиця повинна мати достатній розмір, інакше з’являються проблеми з пошуком вільних комірок
