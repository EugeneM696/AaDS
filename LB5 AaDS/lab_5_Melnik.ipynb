2.Завдання для самостійної роботи: оцінити асимптотичну складність алгоритму бінарного пошуку в О-нотації.

Бінарний пошук має логарифмічну часову складність: O(log n)

3.Завдання для самостійної роботи: написати функцію, яка б могла повертати датафрейм з таким складом полів: («n», «time»), де n - розмір масиву для функції linear_searh();

```python
import time
import random

def linear_search(arr, target):
    for i, x in enumerate(arr):
        if x == target:
            return i
    return -1

def benchmark_linear_search(n_values):
    results = []
    
    for n in n_values:
        arr = list(range(n))
        target = random.choice(arr)

        start = time.perf_counter()
        linear_search(arr, target)
        end = time.perf_counter()
        
        results.append({"n": n, "time": end - start})
    
    return results
    
sizes = [1000, 5000, 10000, 50000, 900000, 9000000, 99999999]
data = benchmark_linear_search(sizes)

for row in data:
    print(row)
```

4.Завдання для самостійної роботи: написати функцію, яка б могла повертати датафрейм з таким складом полів: («n», «time»), де n – розмір масиву для функції bin_searh();

```python
import time
import random

def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

def benchmark_binary_search(n_values):
    results = []
    
    for n in n_values:
        arr = list(range(n))
        target = random.choice(arr)

        start = time.perf_counter()
        binary_search(arr, target)
        end = time.perf_counter()
        
        results.append({"n": n, "time": end - start})
    
    return results
sizes = [1000, 5000, 10000, 50000, 900000, 9000000, 99999999]
data = benchmark_binary_search(sizes)

for row in data:
    print(row)
```

1.Сформулювати задачу пошуку.

Дано:

Цільовий елемент t (елемент, який необхідно знайти).

Потрібно:

Визначити, чи містить структура D елемент t.

3.Що таке лінійний пошук і яка його асимптотична складність?

Лінійний пошук - це простий алгоритм пошуку елемента у колекції даних (масив, список тощо), який працює послідовно.

Найгірший випадок: O(n)

Середній випадок: Θ(n)

Кращий випадок: Ω(1)

4.Що таке бінарний (двійковий) пошук і в чому полягає його головний принцип роботи?

Бінарний пошук – це ефективний алгоритм пошуку елемента у відсортованому масиві. Головний принцип роботи:

Береться середній елемент масиву.

Порівнюється середній елемент з шуканим значенням target:

Якщо середній елемент дорівнює target → пошук завершено.

Якщо середній елемент менший за target → шукаємо у правій половині масиву.

Якщо середній елемент більший за target → шукаємо у лівій половині масиву.

Процес повторюється рекурсивно або ітеративно лише на половині масиву, що залишилась.

5.Які переваги і недоліки використання бінарного пошуку порівняно з лінійним?

Переваги бінарного пошуку:

Швидкість (часова складність)

Ефективність при великих даних

Меньше порівнянь

Недоліки бінарного пошуку:

Потрібно відсортовані дані

Складніший у реалізації

Менш ефективний на малих масивах
