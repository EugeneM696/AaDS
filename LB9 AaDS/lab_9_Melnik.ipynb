Z-функція

```python
def z_func(s):
 Z = [len(s)] + [0] * len(s)
 assert len(s) > 1

 for i in range(1, len(s)):
   if s[i] == s[i-1]:
     Z[1] += 1
   else:
     break 

 r, l = 0, 0
 if Z[1] > 0:
   r, l = Z[1], 1

 for k in range(2, len(s)):
   assert Z[k] == 0
   if k > r:
     for i in range(k, len(s)):
       if s[i] == s[i-k]:
         Z[k] += 1
       else:
         break
     r, l = k + Z[k] - 1, k
   else:
     nbeta = r - k + 1
     Zkp = Z[k - l]
     if nbeta > Zkp:
       Z[k] = Zkp
     else:
       nmatch = 0
       for i in range(r+1, len(s)):
         if s[i] == s[i - k]:
           nmatch += 1
         else:
           break
       l, r = k, r + nmatch
       Z[k] = r - k + 1
 return Z

z_func('babarakamana')
```

Стиснення рядка

```python
def compress_with_z(s):
 z_vec = z_func(s)
 for i in range(1, len(s)):
   if (i + z_vec[i] == len(s)) and (z_vec[i] % i == 0):
     return s[:i]
 else:
   return s
s = "тарамакатарамакатарамака"
print(z_func(s))
print(compress_with_z(s))
```

Контрольні питання:

1.Що таке «префікс-функція» у контексті алгоритмів на рядках, і як вона відрізняється від Z-функції?

Префікс-функція та Z-функція — це два фундаментальні інструменти в алгоритмах на рядках. Обидві вони описують структуру збігів усередині рядка, але роблять це по-різному. 

Для рядка s[0…n−1] префікс-функція — це масив π[i], де: 𝜋[i]=довжина найдовшого власного префікса рядка 𝑠, який є також суфіксом 𝑠[0..𝑖]

Тобто π[i] показує: наскільки префікс рядка «співпадає» із підрядком, що закінчується в позиції i.

Відмінність в тому що префікс-функція сканує зліва направо, контролює довжину поточного збігу в кінці. Z-функція сканує зліва направо, але порівнює підрядок від i з початком рядка.

2.Що таке Z-функція у вищому контексті алгоритмів на рядках, і яка її роль у вирішенні задач?

Z-функція - це інструмент аналізу рядків, який для кожної позиції i вказує, наскільки фрагмент рядка, що починається з i, збігається з його префіксом. У ширшому контексті алгоритмів на рядках Z-функція є одним з ключових способів виявлення повторів, узгоджень і структурних властивостей тексту. Її обчислення можливе за час O(n), що робить її дуже ефективною для великих рядків. Роль і значення Z-функції у задачах на рядках: пошук підрядка за O(n), виявлення повторів і періодів рядка, алгоритми стиснення, порівняння фрагментів рядка за O(1).
