Завдання для самостійної роботи 1:

− написати функцію pop_n(), що видаляє елементи стека з його початку до номера n включно;

− оцінити асимптотичну складність (у середньому і в найгіршому разі) процедур search, insert і delete роботи зі стеком.

```python
def pop_n(stack, n):
    if n >= len(stack):
        removed = stack.copy()
        stack.clear()
        return removed

    removed = stack[:n+1]
    del stack[:n+1]
    return removed

stack = [10, 20, 30, 40, 50]

deleted = pop_n(stack, 0)

print("Видалено:", deleted)
print("Стек після видалення:", stack) 
```

search: Середній випадок-O(n), найгірший випадок-O(n)

insert: Середній випадок-O(1), найгірший випадок-O(1)

delete: Середній випадок-O(1), найгірший випадок-O(1)

Завдання для самостійної роботи 2:

− розглянути самостійно Приклад 3 офіційної документації модуля Queue;

− написати функцію print_n(), що друкує елементи черги з його початку до номера n включно;

− оцінити асимптотичну складність (у середньому і в найгіршому разі) процедур search, insert і delete роботи з чергою.

```python
def print_n(queue, n):

    if n >= len(queue):
        n = len(queue) - 1

    for i in range(0, n + 1):
        print(queue[i])

queue = ["A", "B", "C", "D", "E"]

print_n(queue, 2)
```

search: Середній випадок-O(n), найгірший випадок-O(n)

insert: Середній випадок-O(1), найгірший випадок-O(1)

delete: Середній випадок-O(1), найгірший випадок-O(1)

1.Що таке стек і які операції можна виконувати зі стеком?

   Стек – це базова структура даних, у якій можна тільки вставляти або видаляти елементи на початку стека. Стек організовано за принципом LIFO (Last In First Out) – це означає, що останній елемент, який додано до стеку, – це перший елемент, який з нього виходить.
   
   Операції зі стеком
1. Stack() – створює новий пустий стек. Параметри не потрібні, повертає
пустий стек.
2. push(item) – додає новий елемент на вершину стека. Параметром є
елемент; функція нічого не повертає.
3. pop() – видаляє верхній елемент зі стека. Параметри не потребуються,
функція повертає елемент. Стек змінюється.
4. peek() – повертає верхній елемент стека, але не видаляє його. Параметри
не потребуються, стек не модифікується.
5. isEmpty() – перевіряє стек щодо пустоти. Параметри не потребуються,
повертає бульове значення.
6. size() – повертає кількість елементів у стеку. Параметри не
потребуються, тип результату – ціле число.

3.Як ви можете реалізувати стек за допомогою масиву і за допомогою зв’язаного списку? Які переваги та недоліки кожного підходу?

Реалізація стеку за допомогую масиву:

```python
class StackArray:
    def __init__(self):
        self.stack = []

    def push(self, value):
        self.stack.append(value)

    def pop(self):
        return self.stack.pop() if self.stack else None

    def top(self):
        return self.stack[-1] if self.stack else None

    def is_empty(self):
        return len(self.stack) == 0
```

Переваги: дуже швидкі push і pop, елементи в пам’яті зберігаються компактно, простота реалізації

Недоліки: потрібне перевиділення пам’яті, не можна ефективно видаляти з початку, розмір збільшується ступінчасто

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class StackLinkedList:
    def __init__(self):
        self.top_node = None

    def push(self, value):
        new_node = Node(value)
        new_node.next = self.top_node
        self.top_node = new_node

    def pop(self):
        if not self.top_node:
            return None
        value = self.top_node.value
        self.top_node = self.top_node.next
        return value

    def top(self):
        return self.top_node.value if self.top_node else None

    def is_empty(self):
        return self.top_node is None
```

Переваги: завжди O(1) для push і pop, не потребує великого цілісного блоку пам’яті, може рости, поки є пам’ять

Недоліки: додаткова пам’ять, гірша кеш-локальність, складніша реалізація
