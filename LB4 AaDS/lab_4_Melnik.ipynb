Мета: засвоїти технологію реалізації рекурсивних алгоритмів засобами Python і оцінювання їх складності з використанням головної теореми рекурсії.

Хід роботи:

1.Завдання для самостійної роботи: Реалізувати процедуру обчислення факторіалу за допомогою циклу for. Оцінити асимптотичну складність алгоритму.

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
```

Асимптотична складність алгоритму: O(n)

2.Завдання для самостійної роботи: Оцінити асимптотичну складність рекурсивного алгоритму обчислення факторіалу.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
```

Асимптотична складність рекурсивного обчислення факторіалу - лінійна, 𝑂(𝑛).

3.Завдання для самостійної роботи: Оцінити асимптотичну складність рекурсивного алгоритму обчислення n-го числа Фібоначчі.

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)
```

Асимптотична складність рекурсивного алгоритму Фібоначчі: O(2^n) або 𝑂(1.618^n)

T(n)=O(nlogn) Складність O(nlogn) є оптимальною для сортування на основі порівнянь і є однією з ключових переваг сортування злиттям, оскільки ця складність зберігається як у найгіршому, так і в середньому випадку.
